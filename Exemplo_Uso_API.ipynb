# Exemplo de Uso - Hydrodash API
Neste notebook, demonstramos como realizar a autenticação e extrair dados de previsão de chuva para o Rio Grande do Sul utilizando o endpoint `/previsao_chuva_rs`.

```python
import requests
import pandas as pd
```

### 1. Autenticação
A API utiliza OAuth2. Você precisa enviar suas credenciais para o endpoint `/token`.

```python
BASE_URL = 'https://hydrodash-api.onrender.com'

auth_data = {
    'username': 'SEU_USUARIO',
    'password': 'SUA_SENHA'
}

response = requests.post(f'{BASE_URL}/token', data=auth_data)
if response.status_code == 200:
    token = response.json().get('access_token')
    print('Autenticado com sucesso!')
else:
    print('Erro na autenticação:', response.text)
```

### 2. Consultar Previsão de Chuva (RS)
Passaremos o token no cabeçalho e consultaremos o endpoint passando filtros como `nome`.

```python
headers = {
    'Authorization': f'Bearer {token}'
}

params = {
    'nome': 'Alegre',
    'tipo': 'municipio',
    'limit': 10
}

req_data = requests.get(f'{BASE_URL}/previsao_chuva_rs', headers=headers, params=params)

if req_data.status_code == 200:
    data = req_data.json()
    df = pd.DataFrame(data)
    display(df.head())
else:
    print('Erro na consulta:', req_data.text)
```
